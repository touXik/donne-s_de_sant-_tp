## Étape 2 – Nettoyage et préparation 

```python
df[df.duplicated(subset=['lastname','firstname'],keep=False)]
```

```python
df[(df['firstname']=='Eve') & (df['lastname']=='David')]
```

```python
df[df.duplicated()]
```

```python
plt.figure(figsize=(20,10))
sns.heatmap(df.isna(),cbar=False)
```

```python
df=df.dropna()
```

```python
df[df['baseline value']=='bas']
```

```python
df.loc[df['baseline value']<=110,'baseline classe']='bas'
df.loc[(df['baseline value']>110) & (df['baseline value']<160) ,'baseline classe' ]='moyen'
df.loc[df['baseline value']>=160,'baseline classe']= 'élevé'
```

```python
df
```

```python
df.columns
```

```python
idx = df.columns.get_loc('baseline value') + 1
idx
```

```python
col = df.pop('baseline classe')
df.insert(idx,'baseline classe',col)
```

```python
df
```

```python
df['id']=['ID'+str(i+1).zfill(3) for i in range(len(df))]
    
```

```python
df
```

```python
col = df.pop('id')
```

```python

idx=df.columns.get_loc('firstname')+1
df.insert(idx,'id',col)
```
